```python
import random
from pathlib import Path
from xsteiner.evol.custom import CustomPopulation as Population
from xsteiner.evol.evaluation import (
    evaluate_tree,
    normalize
)
from xsteiner.evol.populate import (
    PrimRSTPopulation,
    KruskalRSTPopulation,
    BoruvkaRSTPolulation,
    RandomWalkRSPPopulation
)
from xsteiner.graph.operators import (
    check_cycle_union_find,
    check_cycle_dfs,
    is_connected,

)
from xsteiner.utils.orlib import orlib_parser
```

```python
FOLDER = Path('datasets')

assert FOLDER.exists()

files = list(FOLDER.glob('*.txt'))
len(files)
```

```python
index = 15
print(files[index])
stpg = orlib_parser(files[index])
```

```python
generator = PrimRSTPopulation(stpg)
# generator = KruskalRSTPopulation(stpg)
# generator = BoruvkaRSTPolulation(stpg)
# generator = RandomWalkRSPPopulation(stpg)
```

```python
population = [generator() for _ in range(200)]

len(population)
```

```python
bool_check_cycle = [not check_cycle_dfs(p, random.choice(list(p.nodes())))
                    for p in population]

print(all(bool_check_cycle))


bool_is_connected = [is_connected(p) for p in population]
print(all(bool_is_connected))
```

```python
cost = [evaluate_tree(p) for p in population]
```

```python
pop = Population(population, eval_function=evaluate_tree)
```
